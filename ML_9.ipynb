```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Load the dataset
df = pd.read_csv('/content/Iris.csv')
print("Dataset Columns:", df.columns.tolist())

# Separate features (excluding ID and species target if present)
feature_cols = [col for col in df.columns if col not in ['Id', 'Species', 'species', 'id']]

# Drop missing values to handle the ValueError
print(f"Original dataset shape: {df.shape}")
df_clean = df.dropna(subset=feature_cols)
print(f"Cleaned dataset shape: {df_clean.shape}")

X = df_clean[feature_cols]

# Standardize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 1. PCA with 2 Components
pca_2 = PCA(n_components=2)
pca_2.fit(X_scaled)
var_2 = pca_2.explained_variance_ratio_

# 2. PCA with 3 Components
pca_3 = PCA(n_components=3)
pca_3.fit(X_scaled)
var_3 = pca_3.explained_variance_ratio_

print(f"\n--- PCA with 2 Components ---")
print(f"Explained Variance per PC: {var_2}")
print(f"Total Explained Variance (Cumulative): {sum(var_2):.4f} ({sum(var_2)*100:.2f}%)")

print(f"\n--- PCA with 3 Components ---")
print(f"Explained Variance per PC: {var_3}")
print(f"Total Explained Variance (Cumulative): {sum(var_3):.4f} ({sum(var_3)*100:.2f}%)")
```
