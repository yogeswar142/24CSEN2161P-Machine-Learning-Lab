### Load the Dataset and Explore Columns
Let's first load the red wine quality dataset and inspect its structure.

```python
import pandas as pd

# Load the dataset
df = pd.read_csv('/content/winequality-red.csv')

# Display basic information and the first few rows
print(df.info())
display(df.head())
```

### Normalization Techniques
We will apply three distinct techniques to the features (excluding the target column `quality` if present, or applying to all features):
1. **Min-Max Scaling**: Scales features to a fixed range (usually 0 to 1).
2. **Standardization (StandardScaler)**: Centers features to have a mean of 0 and a standard deviation of 1.
3. **Robust Scaling**: Scales features using statistics that are robust to outliers (IQR: Interquartile Range).

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler

# Separate features and target (assuming 'quality' is the target to predict in your ML experiment)
if 'quality' in df.columns:
    X = df.drop(columns=['quality'])
    y = df['quality']
else:
    X = df.copy()
    y = None

# 1. Min-Max Scaling
min_max_scaler = MinMaxScaler()
X_minmax = pd.DataFrame(min_max_scaler.fit_transform(X), columns=X.columns)

# 2. Standardization
std_scaler = StandardScaler()
X_std = pd.DataFrame(std_scaler.fit_transform(X), columns=X.columns)

# 3. Robust Scaling
robust_scaler = RobustScaler()
X_robust = pd.DataFrame(robust_scaler.fit_transform(X), columns=X.columns)

# Preview and compare results for a subset of features
features_to_show = list(X.columns[:3])
print("Original features (first 3 rows):")
display(X[features_to_show].head(3))

print("\nMin-Max Scaled features (first 3 rows):")
display(X_minmax[features_to_show].head(3))

print("\nStandardized features (first 3 rows):")
display(X_std[features_to_show].head(3))

print("\nRobust Scaled features (first 3 rows):")
display(X_robust[features_to_show].head(3))
```
